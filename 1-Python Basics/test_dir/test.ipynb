```python
class BankAccount:
    def __init__(self,owner,bal):
        self.owner=owner
        self.bal=bal

    def dep(self,amt):
        self.bal+= amt
        print(f'Amount:{amt} has been deposited updated  and bal is:{self.bal}')

    def wit(self,amt1):
        if amt1 < self.bal:
            self.bal-= amt1
            print(f'Amount {amt1} is withdrawn and Updated bal is {self.bal}')

        else:
            print('Not sufficient balance')

hdfccs1=BankAccount('Arjun',250000)
hdfccs1.dep(200)

hdfccs1.wit(200)
print(hdfccs1.bal)
```
